```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/Users/eldanizarifzada/Downloads/heart_cleveland_upload.csv')
```

```python
df
```

```python
X=df.drop('condition', axis=1)
y=df['condition']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.1)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
def printMetrics(y_pred, y_actual):
    print('Accuracy is ' + str(accuracy_score(y_pred, y_actual)))
    print('Confusion Matrix')
    print(confusion_matrix(y_pred, y_actual))
    print('Classification report')
    print(classification_report(y_pred, y_actual))
```

```python
dt=DecisionTreeClassifier()
dt.fit(X_train, y_train)
pred_dt=dt.predict(X_test)

printMetrics(pred_dt, y_test)
```

```python
rf=RandomForestClassifier()
rf.fit(X_train, y_train)
pred_rf=rf.predict(X_test)

printMetrics(pred_rf, y_test)
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
param_grid = {
    'criterion' : ['gini', 'entropy'],
    'max_depth': [None, 2, 4, 6, 8, 10, 15, 5, 7],
    'min_samples_split' : [2, 5, 10, 3, 4, 7],
    'min_samples_leaf' : [1, 2, 4, 3, 10, 15]
}

dt=DecisionTreeClassifier()
grid_search=GridSearchCV(estimator=dt, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1, verbose=3)
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
grid_search.best_score_
```

```python
new_model= DecisionTreeClassifier(**grid_search.best_params_)
new_model.fit(X_train, y_train)
```

```python
grid_prediction= grid_search.predict(X_test)

printMetrics(grid_prediction, y_test)
```
